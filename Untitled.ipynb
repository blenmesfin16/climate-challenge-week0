```python
# Load the data
df = pd.read_csv('data/ethiopia.csv')

# Add country column
df['Country'] = 'Ethiopia'

print(f"Loaded {len(df)} rows")
print(f"Columns: {df.columns.tolist()}")
df.head()
```

```python
# Replace -999 with NaN (missing)
df = df.replace(-999, np.nan)

# Check missing values
missing_count = df.isna().sum()
missing_percent = (missing_count / len(df)) * 100
missing_df = pd.DataFrame({'Count': missing_count, 'Percent': missing_percent})
missing_df[missing_df['Count'] > 0]
```

```python
# Convert YEAR and DOY to actual date
df['Date'] = pd.to_datetime(df['YEAR'] * 1000 + df['DOY'], format='%Y%j')

# Extract month and year
df['Month'] = df['Date'].dt.month
df['Year'] = df['Date'].dt.year

print(f"Date range: {df['Date'].min()} to {df['Date'].max()}")
df[['YEAR', 'DOY', 'Date', 'Month']].head()
```

```python
# Check for and remove duplicates
duplicates = df.duplicated().sum()
print(f"Duplicate rows found: {duplicates}")

if duplicates > 0:
    df = df.drop_duplicates()
    print(f"Removed {duplicates} duplicates")
```

```python
# Create monthly average temperature
monthly_temp = df.groupby(['Year', 'Month'])['T2M'].mean().reset_index()
monthly_temp['Date'] = pd.to_datetime(monthly_temp['Year'].astype(str) + '-' + monthly_temp['Month'].astype(str))

# Plot
plt.figure(figsize=(14, 6))
plt.plot(monthly_temp['Date'], monthly_temp['T2M'], linewidth=2, color='red')

# Find warmest and coolest
warmest = monthly_temp.loc[monthly_temp['T2M'].idxmax()]
coolest = monthly_temp.loc[monthly_temp['T2M'].idxmin()]

plt.annotate(f'Warmest: {warmest["T2M"]:.1f}°C', 
             xy=(warmest['Date'], warmest['T2M']),
             xytext=(10, 5), textcoords='offset points')
plt.annotate(f'Coolest: {coolest["T2M"]:.1f}°C',
             xy=(coolest['Date'], coolest['T2M']),
             xytext=(10, -15), textcoords='offset points')

plt.title('Ethiopia Monthly Average Temperature (2015-2026)')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Save cleaned data for later use
df.to_csv('data/ethiopia_clean.csv', index=False)
print("Saved cleaned Ethiopia data to data/ethiopia_clean.csv")
```

```python
# Monthly total precipitation
monthly_rain = df.groupby(['Year', 'Month'])['PRECTOTCORR'].sum().reset_index()
monthly_rain['Date'] = pd.to_datetime(monthly_rain['Year'].astype(str) + '-' + monthly_rain['Month'].astype(str))

# Find rainy season
monthly_avg_rain = df.groupby('Month')['PRECTOTCORR'].mean()
peak_months = monthly_avg_rain.nlargest(3).index.tolist()
print(f"Peak rainy months: {peak_months}")

# Plot
plt.figure(figsize=(14, 6))
plt.bar(monthly_rain['Date'], monthly_rain['PRECTOTCORR'], color='blue', alpha=0.7)
plt.title('Ethiopia Monthly Total Precipitation (2015-2026)')
plt.xlabel('Date')
plt.ylabel('Precipitation (mm)')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Select numeric columns for correlation
corr_cols = ['T2M', 'T2M_MAX', 'T2M_MIN', 'PRECTOTCORR', 'RH2M', 'WS2M']
corr_matrix = df[corr_cols].corr()

# Plot heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')
plt.title('Correlation Between Climate Variables - Ethiopia')
plt.show()

# Find strongest correlations
print("\nStrongest correlations:")
upper = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
strongest = upper.unstack().abs().sort_values(ascending=False).dropna()
print(strongest.head(3))
```

```python
# Histogram of precipitation
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Regular scale
ax1.hist(df['PRECTOTCORR'], bins=50, edgecolor='black', color='skyblue')
ax1.set_title('Precipitation Distribution (Linear)')
ax1.set_xlabel('Precipitation (mm/day)')
ax1.set_ylabel('Frequency')

# Log scale (better for skewed data)
ax2.hist(df['PRECTOTCORR'] + 0.01, bins=50, log=True, edgecolor='black', color='skyblue')
ax2.set_title('Precipitation Distribution (Log Scale)')
ax2.set_xlabel('Precipitation (mm/day)')
ax2.set_ylabel('Frequency (log scale)')

plt.tight_layout()
plt.show()

print("Note: Most days have little rain, few days have heavy rain - typical for seasonal climate")
```

```python
# Process all 5 countries
countries = ['ethiopia', 'kenya', 'sudan', 'tanzania', 'nigeria']
all_data = []

for country in countries:
    # Load
    temp_df = pd.read_csv(f'data/{country}.csv')
    temp_df['Country'] = country.capitalize()
    
    # Clean
    temp_df = temp_df.replace(-999, np.nan)
    temp_df = temp_df.drop_duplicates()
    
    # Parse dates
    temp_df['Date'] = pd.to_datetime(temp_df['YEAR'] * 1000 + temp_df['DOY'], format='%Y%j')
    temp_df['Month'] = temp_df['Date'].dt.month
    temp_df['Year'] = temp_df['Date'].dt.year
    
    # Fill missing
    temp_df = temp_df.fillna(method='ffill')
    temp_df = temp_df.fillna(temp_df.median())
    
    all_data.append(temp_df)
    print(f"Processed {country}: {len(temp_df)} rows")

# Combine all
df_all = pd.concat(all_data, ignore_index=True)
print(f"\nTotal combined data: {len(df_all)} rows")
```

```python
# Process all 5 countries (simpler version)
countries = ['ethiopia', 'kenya', 'sudan', 'tanzania', 'nigeria']
all_data = []

for country in countries:
    print(f"Processing {country}...")
    
    # Load
    temp_df = pd.read_csv(f'data/{country}.csv')
    temp_df['Country'] = country.capitalize()
    
    # Replace -999 with NaN
    temp_df = temp_df.replace(-999, np.nan)
    
    # Remove duplicates
    temp_df = temp_df.drop_duplicates()
    
    # Parse dates
    temp_df['Date'] = pd.to_datetime(temp_df['YEAR'] * 1000 + temp_df['DOY'], format='%Y%j')
    temp_df['Month'] = temp_df['Date'].dt.month
    temp_df['Year'] = temp_df['Date'].dt.year
    
    # Fill missing values (simpler method)
    temp_df = temp_df.fillna(temp_df.mean())
    
    all_data.append(temp_df)
    print(f"  Done! {len(temp_df)} rows")

# Combine all
df_all = pd.concat(all_data, ignore_index=True)
print(f"\n Total combined data: {len(df_all)} rows")
print(f" Countries: {df_all['Country'].unique()}")
```

```python
# Process all 5 countries (with type safety)
countries = ['ethiopia', 'kenya', 'sudan', 'tanzania', 'nigeria']
all_data = []

for country in countries:
    print(f"Processing {country}...")
    
    # Load
    temp_df = pd.read_csv(f'data/{country}.csv')
    temp_df['Country'] = country.capitalize()
    
    # Convert numeric columns to numbers (errors='coerce' turns bad values into NaN)
    numeric_cols = ['YEAR', 'DOY', 'T2M', 'T2M_MAX', 'T2M_MIN', 'T2M_RANGE', 
                    'PRECTOTCORR', 'RH2M', 'WS2M', 'WS2M_MAX', 'PS', 'QV2M']
    
    for col in numeric_cols:
        if col in temp_df.columns:
            temp_df[col] = pd.to_numeric(temp_df[col], errors='coerce')
    
    # Replace -999 with NaN
    temp_df = temp_df.replace(-999, np.nan)
    
    # Remove duplicates
    temp_df = temp_df.drop_duplicates()
    
    # Parse dates (convert to numbers first)
    temp_df['YEAR'] = pd.to_numeric(temp_df['YEAR'], errors='coerce')
    temp_df['DOY'] = pd.to_numeric(temp_df['DOY'], errors='coerce')
    
    # Remove rows with invalid dates
    temp_df = temp_df.dropna(subset=['YEAR', 'DOY'])
    
    # Create date
    temp_df['Date'] = pd.to_datetime(temp_df['YEAR'].astype(int) * 1000 + temp_df['DOY'].astype(int), 
                                      format='%Y%j', errors='coerce')
    temp_df['Month'] = temp_df['Date'].dt.month
    temp_df['Year'] = temp_df['Date'].dt.year
    
    # Remove rows with invalid dates
    temp_df = temp_df.dropna(subset=['Date'])
    
    # Fill missing values with column mean
    temp_df = temp_df.fillna(temp_df.mean(numeric_only=True))
    
    all_data.append(temp_df)
    print(f"  Done! {len(temp_df)} rows")

# Combine all
df_all = pd.concat(all_data, ignore_index=True)
print(f"\n Total combined data: {len(df_all)} rows")
print(f" Countries: {df_all['Country'].unique().tolist()}")
```

```python
# Monthly averages per country
monthly_all = df_all.groupby(['Country', 'Year', 'Month'])['T2M'].mean().reset_index()
monthly_all['Date'] = pd.to_datetime(monthly_all['Year'].astype(str) + '-' + monthly_all['Month'].astype(str))

# Plot
plt.figure(figsize=(14, 7))
for country in monthly_all['Country'].unique():
    country_data = monthly_all[monthly_all['Country'] == country]
    plt.plot(country_data['Date'], country_data['T2M'], label=country, linewidth=2)

plt.title('Temperature Comparison Across 5 African Countries (2015-2026)')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Summary table
temp_summary = df_all.groupby('Country')['T2M'].agg(['mean', 'median', 'std']).round(2)
temp_summary.columns = ['Mean (°C)', 'Median (°C)', 'Std Dev']
print("\n📊 Temperature Summary by Country:")
print(temp_summary)
```

```python
# Boxplot comparison
plt.figure(figsize=(12, 6))
sns.boxplot(data=df_all, x='Country', y='PRECTOTCORR')
plt.yscale('log')
plt.title('Precipitation Distribution by Country')
plt.ylabel('Precipitation (mm/day) - Log Scale')
plt.xticks(rotation=45)
plt.grid(True, alpha=0.3)
plt.show()

# Summary table
precip_summary = df_all.groupby('Country')['PRECTOTCORR'].agg(['mean', 'median', 'std']).round(2)
precip_summary.columns = ['Mean (mm)', 'Median (mm)', 'Std Dev']
print("\n📊 Precipitation Summary by Country:")
print(precip_summary)
```

```python
# Extreme heat days (T2M_MAX > 35°C)
df_all['Extreme_Heat'] = df_all['T2M_MAX'] > 35

# Calculate per country per year
extreme_by_country = df_all.groupby(['Country', 'Year'])['Extreme_Heat'].sum().reset_index()

# Plot
plt.figure(figsize=(12, 6))
for country in extreme_by_country['Country'].unique():
    country_data = extreme_by_country[extreme_by_country['Country'] == country]
    plt.plot(country_data['Year'], country_data['Extreme_Heat'], marker='o', label=country, linewidth=2)

plt.title('Days with Temperature Above 35°C per Year')
plt.xlabel('Year')
plt.ylabel('Number of Extreme Heat Days')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Summary
extreme_summary = extreme_by_country.groupby('Country')['Extreme_Heat'].mean().round(1)
print("\n🔥 Average Extreme Heat Days per Year:")
print(extreme_summary.sort_values(ascending=False))
```

```python
# Calculate vulnerability metrics
vulnerability_data = []

for country in df_all['Country'].unique():
    country_df = df_all[df_all['Country'] == country]
    
    # Temperature trend (warming rate)
    yearly_temp = country_df.groupby('Year')['T2M'].mean()
    years = np.arange(len(yearly_temp))
    slope, _ = np.polyfit(years, yearly_temp.values, 1)
    
    # Extreme heat days per year
    extreme_days = extreme_by_country[extreme_by_country['Country'] == country]['Extreme_Heat'].mean()
    
    # Precipitation variability
    precip_std = country_df['PRECTOTCORR'].std()
    
    # Mean temperature
    mean_temp = country_df['T2M'].mean()
    
    vulnerability_data.append({
        'Country': country,
        'Mean Temp (°C)': round(mean_temp, 1),
        'Warming Rate (°C/year)': round(slope, 4),
        'Extreme Heat Days/Year': round(extreme_days, 1),
        'Precip Variability': round(precip_std, 2)
    })

vuln_df = pd.DataFrame(vulnerability_data)

# Calculate vulnerability score (higher = more vulnerable)
vuln_df['Vulnerability Score'] = (
    vuln_df['Warming Rate (°C/year)'].rank(ascending=False) +
    vuln_df['Extreme Heat Days/Year'].rank(ascending=False) +
    vuln_df['Precip Variability'].rank(ascending=False)
)

print("\n" + "="*70)
print("🌍 CLIMATE VULNERABILITY RANKING FOR COP32 POSITION PAPER")
print("="*70)
print(vuln_df.sort_values('Vulnerability Score')[['Country', 'Mean Temp (°C)', 'Warming Rate (°C/year)', 
                                                    'Extreme Heat Days/Year', 'Precip Variability', 'Vulnerability Score']])

print("\n" + "="*70)
print("🏆 RANKING (Most Vulnerable to Least Vulnerable):")
ranking_order = vuln_df.sort_values('Vulnerability Score')['Country'].tolist()
for i, country in enumerate(ranking_order, 1):
    print(f"   {i}. {country}")
```

```python
print("\n" + "="*70)
print("📋 5 EVIDENCE-BACKED OBSERVATIONS FOR COP32")
print("="*70)

# Get top vulnerable country
most_vulnerable = ranking_order[0]
warming_fastest = vuln_df.loc[vuln_df['Warming Rate (°C/year)'].idxmax(), 'Country']
most_extreme = vuln_df.loc[vuln_df['Extreme Heat Days/Year'].idxmax(), 'Country']

observations = f"""
1. WHAT IS CHANGING?
   {warming_fastest} is warming fastest at {vuln_df['Warming Rate (°C/year)'].max():.4f}°C per year,
   with rising extreme heat days affecting agricultural planning.

2. WHAT HAS IT CAUSED?
   {most_vulnerable} shows {vuln_df.loc[vuln_df['Country']==most_vulnerable, 'Extreme Heat Days/Year'].values[0]} extreme heat days annually,
   reducing labor productivity and increasing cooling demands.

3. WHAT DOES IT DEMAND?
   Climate finance for early warning systems and heat-resilient infrastructure in
   {most_vulnerable} and {warming_fastest}.

4. HOW DOES ETHIOPIA COMPARE?
   Ethiopia shows moderate warming but significant precipitation variability,
   requiring investment in water storage and drought-resistant crops.

5. WHO SHOULD GET PRIORITY FINANCE?
   {most_vulnerable} should be championed for priority climate adaptation finance,
   supported by data showing highest vulnerability score of {vuln_df['Vulnerability Score'].max():.0f}.
"""

print(observations)
```

```python
# Save vulnerability ranking
vuln_df.to_csv('vulnerability_ranking.csv', index=False)
print("✅ Saved vulnerability_ranking.csv")

# Save combined cleaned data
df_all.to_csv('data/all_countries_clean.csv', index=False)
print("✅ Saved all_countries_clean.csv")

# Save observations to text file
with open('cop32_observations.txt', 'w') as f:
    f.write(observations)
print("✅ Saved cop32_observations.txt")

print("\n🎉 All analysis complete! Ready for final report!")
```
